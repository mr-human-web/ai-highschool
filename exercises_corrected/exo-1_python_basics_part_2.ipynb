<div class="alert alert-block alert-info">
<b>Contexte:</b>

Dans la dernière série d'exercices, vous avez pu voir comment calculer le temps de vol d'un objet en chute libre lancé directement vers le haut. Je vous montre ici la dérivation pour un objet lancé 
à un certain angle vers le haut par rapport à l'horizontale (voir l'esquisse ci-dessous)

Après avoir appliqué la seconde loi de Newton, on obtient les composantes suivantes à l'accélération $\vec{a}$
$$
\begin{cases}
a_x(t) &= 0 
\\
a_y(t) &= - g
\end{cases}
$$
Après avoir intégré, on obtient 
$$
\begin{cases}
v_x(t) &= v_0\cos(\theta)
\\
v_y(t) &=  - g t + v_0\sin(\theta)
\end{cases}
$$
où $\theta$ est l'angle que fait le vecteur vitesse $\vec{\mathbf{v}}$ avec l'horizontale.  

Finalement, après avoir intégré une seconde fois, on obtient le vecteur position $\vec{\mathbf{r}}=[x(t),y(t)]^{\top}$,
$$
\begin{cases}
x(t) &= v_0\cos(\theta) t
\\
y(t) &= -\frac{1}{2}gt^2 + v_0\sin(\theta) t
\end{cases}
$$

Il y a une astuce simple à partir d'ici pour trouver l'angle optimal pour notre pauvre lanceur de poids mais pour l'instant je vous laisse réfléchir un peu sur la méthode. En attendant, continuons la dérivation comme la dernière fois. 

Le temps de vol $T$ est obtenu en trouvant les racines d'une équation du second degré, celles de $y(t)$. Ici, la résolution est plus simple puisqu'on assume que $y(0)=0$, donc $y(T)=0$ est obtenu en isolant $T$ dans l'équation suivante
$$
\begin{align}
y(T) &= -\frac{1}{2}gT +v_0\sin(\theta)&=0 \Longrightarrow T =\frac{2v_0\sin(\theta)}{g} 
\end{align}
$$
En remplaçant $T$ dans l'équation de $x(t)$, on obtient la portée, soit la longueur parcourue par le poids. 
$$
\begin{align}
P=x(T)&= v_0\cos(\theta)T
\\
&=v_0\cos(\theta) \times \frac{2 v_0\sin(\theta)}{g}
\\
&=\frac{v_0^2}{g}\times 2\sin(\theta)\cos(\theta)
\\
&= \frac{v_0^2\sin(2\theta)}{g} &&[2\sin(\theta)\cos(\theta) = \sin(2\theta)]
\end{align}
$$

<div class="alert alert-block alert-danger">
<b>Objectif:</b> 

Nous souhaitons créer une fonction qui va calculer la portée, c'est-à-dire la distance en x que parcourra le poids après avoir été lancé à un certain angle par rapport à l'horizontale. Nous souhaiterons ensuite estimer quel angle (en degré) maximise la portée du poids.
</div>

*Question 1: Créer la fonction* `compute_range` 

Votre fonction `compute_range` devra prendre en entrée comme argument l'angle $\theta$ via un paramètre `angle` et devra avoir comme paramètres par défaut la vitesse initiale de la balle `v_init=15` et l'accélération gravitationnelle $g$ que vous noterez `g=9.81`.

Quelques conseils:
- N'oubliez pas de convertir votre angle en radians avant de le passer à la fonction `sin` du package `math`. Cette fonction s'attend à recevoir un angle en radians et pas en degrés. Voici la règle de conversion,
$$
\theta\text{[rad]} = \frac{\pi}{180} \theta\text{[deg]}\, .
$$

````python
from math import sin, pi

def compute_range(angle, v_init = 15, g = 9.81): 
    angle_rad = pi / 180 * angle
    x_range = v_init ** 2 * sin(2 * angle_rad) / g
    return x_range 

compute_range(15)
````

*Question 3: Méthode de recherche de maximum.*

Dans la question précédente, on a utilisé ce qu'on appelle une *grid search* (ou recherche sur maille en bon français). Cette méthode consiste à chercher la valeur qui maximise une fonction (dans notre cas, la portée en fonction de l'angle) en faisant plusieurs essais-erreurs. Le terme "grid" (pour maille) vient du fait qu'on a maillé un espace continu: on a créé une liste d'angles contenant des valeurs discrètes d'angles entre 10 et 90 degrés. 

La *grid search* est une méthode qui est notamment utile en grandes dimensions (comme on le verra avec les réseaux de neurones lorsqu'on parlera d'_hyperparameter tuning_). Ici, comme on a qu'une seule dimension, je voulais vous montrer un autre algorithme qui permet de trouver un maximum d'une fonction qui n'est pas trop bizarre. 

L'algorithme est le suivant:

- Étape 1: Proposer deux points candidats $x_0$ et $x_1$ qui encadrent le maximum;

- Étape 3: Sélectionner la zone contenant le maximum. Ceci peut être fait en suivant les sous-étapes suivantes:

(i) Calculer le triplet de valeurs $[f(x_0), f(x_{1/2}), f(x_1)]$ 

(ii) Tester les inégalités suivantes $f(x_0)<f(x_{1/2})$ et $f(x_1)<f(x_{1/2})$. Si une de ces deux inégalités est satisfaite, assigner $x_{1/2}$ à la variable $x^{\text{next}}_{0}$. Sinon, l'algorithme échoue, il faut resélectionner les points candidats. 

(iii) Tester l'inégalité suivante $f(x_0)<f(x_1)$. Si cette inégalité est satisfaite, assigner $x_1$ à la variable $x^{\text{next}}_{1}$. Sinon, assigner $x_0$ à la variable $x^{\text{next}}_{1}$.

(iv) Calculer un critère d'arrêt de l'algorithme $\epsilon=\lvert f(x_0)-f(x^\text{next}_0)\rvert$. Finalement, assigner $x_0 = x_0^{\text{next}}$ et $x_{1} = x_{1}^{\text{next}}$ pour la prochaine itération de l'algorithme. La fonction `abs` qui calcule une valeur absolue et la fonction `max` vous sera utile.

- Étape 4: Répéter les étapes 2 et 3 itérativement jusqu'à ce que le critère d'arrêt $\epsilon$ devienne plus petit qu'une valeur seuil $\epsilon_{\min}$. Pour notre cas, mettons $\epsilon_{\min}=10^{-3}$.

**Votre tâche (si vous l'acceptez) est d'implémenter l'algorithme décrit ci-dessus pour la fonction `compute_range` (fonction de l'angle $\theta$) avec $x_0=\theta_0=10$ degrés et $x_1=\theta_1=90$ degrés. Affichez les points finaux $x_0$, $x_1$ et $x_{1/2}$. Que déterminez-vous sur l'angle final maximisant la portée du tir du lanceur de poids ?**

*Indice 1:* Utilisez une boucle `while` pour répéter les étapes 2 et 3 de l'algorithme.

*Indice 2:* Vous avez besoin de définir les variables suivantes au début de votre code avant de commencer la boucle `while`: $x_0$, $x_1$, $\epsilon_{\min}$, $x_{1/2}$.

````python
# Step 1: Point candidates
x0 = 10
x1 = 100

# Step 2
x_mid = (x0 + x1) / 2

eps_min = 1e1

i = 0


# Step 3 and 4
while eps_min >1e-8:
    f0 = compute_range(x0)
    fmid = compute_range(x_mid)
    f1 = compute_range(x1)

    if f0 < fmid:
        x0_next = x_mid

    elif f1 < fmid:
        x0_next = x_mid 

    if f0 < f1:
        x1_next = x1
    else:
        x1_next = x0
    
    x_mid_next = (x0_next + x1_next) / 2
    eps_min = abs(compute_range(x_mid) - compute_range(x_mid_next)) 

    # Replacing the variables for the next loop of the algorithm and Step 2 again
    x0 = x0_next
    x1 = x1_next
    x_mid = x_mid_next

print(x_mid)
````

Pour terminer cette exercice, je vous montre comment déterminer analytiquement que la portée est maximisée pour un angle d'exactement 45 degrés. Je vous montre deux méthodes, une utilisant les propriétés de la fonction sinus tandis que l'autre utilise la dérivée et les propriétés de la fonction cosinus.


**Méthode 1: Propriétés de la fonction sinus.**

La portée, en tant que fonction de l'angle $\theta$ est donnée par
$$
\begin{align}
x(\theta) = \frac{v_0^{2}\sin(2\theta)}{g}
\end{align}
$$

La fonction $\sin(x)$ est bornée entre [-1, 1]. Les maxima de la fonction $\sin$ sont obtenus périodiquement à 
$$
x_{\text{max}}=\pi/2 + 2k\pi \quad \forall k\in \mathbb{Z}
$$ 
où $\mathbb{Z}$ est l'espace des entiers relatifs. 

Posant $x=2\theta$, on obtient 
$$
\theta_{\max}=\frac{x_{\max}}{2} = \frac{\pi}{4} + k\pi \quad \forall k \in \mathbb{Z}
$$. 

Assumant qu'on s'intéresse seulement aux angles appartenant à l'intervalle $[0, \frac{\pi}{2}]$, on déduit que le seul angle maximisant la fonction $\sin$ dans cet intervalle est $\theta_{\max}=\frac{\pi}{4}$ ce qui correspond à un angle de 45 degrés.  

**Méthode 2: La dérivée.**

Cette seconde méthode a le bénéfice de fonctionner pour n'importe quelle fonction. Donc à privilégier si la fonction qu'on étudie est plutôt complexe. 

Un extremum d'une fonction est atteint lorsque la dérivée est nulle. L'extremum peut être
1) Un maximum 
2) Un minimum
3) Un point de selle. 

Lorsqu'on cherche un maximum, il faut que la dérivée soit nulle et une deuxième condition doit être remplie; il faut que la dérivée seconde à l'extremum soit négative. En maths, ça donne,
$$
\frac{d^2f}{dx^2}(x_{\text{extremum}})<0 \quad \text{ou} \quad f''(x_{\text{extremum}})<0 \, .
$$

Allons-y avec tout ça pour la dérivation. La fonction d'intérêt est $x(\theta)$. 

(i) On commence par calculer la dérivée
$$
\begin{align}
x'(\theta) = \frac{d x}{d\theta} &=  \frac{2v_0^{2}\cos(2\theta)}{g} 
\end{align}
$$

(ii) On calcule le point $\theta_{\text{extremum}}$ où la dérivée est égale à 0,
$$
\begin{align}
    x'(\theta) = \frac{2v_0^{2}\cos(2\theta)}{g}-0 \Longrightarrow \cos(2\theta) = 0
\end{align}
$$

Bon... il faut faire maintenant appel aux propriétés des fonctions trigonométriques pour résoudre cette équation et pour montrer que la dérivée seconde est inférieure à 0... Ça sent le déjà-vu qui arrive... Vous comprenez maintenant pourquoi j'ai fait un petit aparté sur la méthode 1 ;).

Les zéros de la fonction cosinus sont obtenus aux point suivants, 
$$
\begin{align}
\cos(x) = 0 \Longrightarrow x = \frac{\pi}{2} +k\pi \quad \forall k\in \mathbb{Z}
\end{align}
$$

On pose $x=2\theta$ et on obtient les zéros suivants pour la fonction $\cos(2\theta)$,
$$
\begin{align}
\cos(2\theta) = 0 \Longrightarrow 2\theta &= \frac{\pi}{2} +k\pi \quad \forall k \in \mathbb{Z}
\\
\theta&=\frac{\pi}{4} + k\frac{\pi}{2} \quad \forall k\in \mathbb{Z}
\end{align}
$$
On peut encore utiliser l'argument qu'on s'intéresse seulement à des angles appartenant à l'intervalle $[0, \frac{\pi}{2}]$ pour juste s'íntéresser à $\theta_{\text{extremum}}=\frac{\pi}{4}$. 

Il faut finalement montrer que notre extremum est un maximum.
