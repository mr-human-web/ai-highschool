##### Exercice 0.0: Chute libre d'une balle
Une balle est lancée à la verticale, vers le haut, avec une vitesse initiale $v_0$. Nathan veut utiliser l'équation d'un objet en chute libre pour déterminer après combien de temps la balle retombera au sol (c.à.d trouver l'instant $T$ où $y=0$ m).

L'équation d'un objet en chute libre est donnée par, 
$$
y = -\frac{1}{2}gt^2 + v_0 t + y_0 \, .
$$

Nathan déduit que la balle retombera au sol après une durée de temps $T$,
$$
-\frac{1}{2}gT^2 +v_0T +y_0 = 0
$$

Cette équation quadratique peut être résolue en utilisant la formule permettant d'en calculer les racines,
$$
\begin{align}
T_{\pm} &= \frac{-b\pm \sqrt{b^2 - 4ac}}{2 a}
\end{align}
$$
où
$$
\begin{align}
a &= -\frac{1}{2}g
\\
b &= v_0
\\
c&= y_0
\end{align} 
$$
On déduit donc que 
$$
\begin{align}
T_\pm &= \frac{- v_0\pm \sqrt{v_0^2+2gy_0}}{-g}
\end{align}
$$

Ici, on s'intéresse seulement à la solution positive. On a donc, 

$$
T = \frac{v_0 + \sqrt{v_0^2 + 2 g y_0}}{g}
$$

Calculez $T$ en utilisant Python avec $v_0=10\text{ m}/\text{s}$, $g=9.81 \text{ m}/\text{s}^2$ et $y_0=10
\text{ m}$.
