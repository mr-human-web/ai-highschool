Tutorial route on how to train models the right way:
1) Defining clearly the task we want to accomplish. In other words, setting the problem. 
2) Dataset preprocessing + visualization. Making sure the dataset is how we expect it to be to not have bad surprises (and bugs). 
3) Creating/Importing the architecture of the neural network. What type of neural network for the task ? Convolutionnal, Recurrent, Attention or not ? Fourier basis trick or not ?
4) Defining the loss objective mathematically and implement it. The loss objective is entirely dependent on the task we want to accomplish  


Usually, the way to go is to do the task above on a simpler problem and then we increase the difficulty to tackle a bigger problem. Doing so helps making a diagnosis of potential issues and to better understand the problem and its structure.
