### Les instructions `if`, `elif` et `else` 
Les instructions `if`, `elif` et `else` servent à exécuter différents blocs de code selon si une condition est satisfaite ou non. 

L'instruction `if` doit être suivie d'une valeur booléenne (souvent un test fait via un opérateur relationnel). Faisons un exemple en code. 

Par exemple, on peut demander à un élève d'entrer sa note à un examen puis tester si cette note est supérieur à 10. En code, ça donne:
```python
grade = input("Entrez votre note").upper()
if grade == "A":
    print("Bravo !")
```
Décrivons ce qui se passe ci-dessus en mots. La fonction `input` permet d'afficher une fenêtre pour l'utilisateur. 

Une fois que l'utilisateur a rempli la fenêtre d'affichage et appuyé sur la touche "Enter", la fenêtre se ferme et la note entrée est stockée dans la variable `grade`. La méthode `.upper()` appliquée sur une variable de type string met en capitales toutes les lettres contenues dans la variable (e.g. `"test".upper()` donne `"TEST"`)

Ensuite, l'instruction `if` associé à l'opération relationnelle `grade.upper()=="A"` permet de tester si la note entrée est égale à A. Si c'est le cas, le message `"Bravo !"` est affiché à l'utilisateur. 

Testons tout ça avec la cellule de code ci-dessous.

````python
grade = input("Entrez votre note").upper()
if grade == "A":
   print("Bravo !")
````

````python
print("test".upper())
````

Maintenant, que faire si l'élève entre la note "B" ? Notre code pour l'instant. Ne vous inquiétez pas l'instruction `elif` vient à la rescousse. Le `elif` est le diminutif de `else` et `if` pour dire "Ou si". Une instruction `elif` doit forcément être utilisée à la suite d'une instruction `if`. En termes de syntaxe, l'utilisation est la même que pour l'instruction `if`.

Dans le cas de notre exemple, l'instruction `elif` va nous permettre d'ajouter plusieurs autres possibilités à notre code, par exemple si l'élève a eu la note "B" ou "C" on peut changer le message à envoyer à l'étudiant. En code, ça va donner quelque chose comme ce qui suit

```python
grade = input("Entrez votre note").upper()
if grade == "A":
    print("Bravo, vous avez eu A! Félicitations !")
elif grade == "B":
    print("Bravo, vous avez eu B! Bien joué !")
elif grade == "C":
    print("Bravo, vous avez eu C! Bien joué !")
elif grade == "D":
    print("Bravo, vous avez eu D! Bien joué !")
```

````python
grade = input("Entrez votre note").upper()
if grade == "A":
    print("Bravo, vous avez eu un A! Félicitations !")
elif grade == "B":
    print("Bravo, vous avez eu un B! Bien joué !")
elif grade == "C":
    print("Bravo, vous avez eu un C! Bien joué !")
elif grade == "D":
    print("Bravo, vous avez eu un D! Bien joué !")
````

Finalement, l'instruction `else` va correspondre au "Sinon..." pour exécuter un bloc de codes si les conditions du `if` ou du/des `elif` précédents ne sont pas respectées.

Pour résumer:
- `if` = "Si la condition 1 est respectée, exécute ce bloc de code"
- `elif`  = "Ou si la condition 2 est respectée, exécute ce bloc de code"
- Ajouter des `elif` au besoin pour plus de conditions.
- `else` = "Sinon, exécute ce bloc de code."

Dans le cadre de notre exemple, le bloc de code du `else` peut correspondre à une erreur où l'utilisateur a entré une note qui n'existe pas (un Z ou un G par exemple). On pourrait donc avoir quelque chose comme ça pour finir notre code
```python
...
else:
    print("Erreur - la note que vous avez entrée n'est pas valide !")
```

````python
grade = input("Entrez votre note").upper()
if grade == "A":
    print("Bravo, vous avez eu un A! Félicitations !")
elif grade == "B":
    print("Bravo, vous avez eu un B! Bien joué !")
elif grade == "C":
    print("Bravo, vous avez eu un C! Bien joué !")
elif grade == "D":
    print("Bravo, vous avez eu un D! Bien joué !")
else:
    print("Erreur - la note que vous avez entrée n'est pas valide !")
````

Ceci conclue l'exemple pratique sur les utilisations des instructions `if`, `elif`, et `else`.  

### Boucles: `for` et `while`

Les boucles permettent de faire des itérations, c'est-à-dire de répéter un bloc de code un certain nombre de fois (dans le cas de la boucle `for`) ou indéfiniment jusqu'à ce qu'une condition ne soit plus satisfaite (dans le cas de la boucle `while`). 

Voyons ces deux instructions plus en détails.

La boucle `for` va notamment être utile pour itérer sur les éléments d'une séquence comme une liste par exemple. On peut faire,
```python
var_list = ["apple", "banana", "pear"]

for fruit in var_list:
    print(fruit)
```

Ici, la boucle `for` permet d'accéder aux différents éléments de la liste `var_list`.
La première itération, le programme accède au `str`, `"apple"`.
La seconde itération, le programme accède au `str`, `"banana"` et ainsi de suite... 

````python
var_list = ["apple", "banana", "pear"]

for fruit in var_list:
    print(fruit)
````

Pour itérer sur des suites arithmétiques, on pourra utiliser la fonction `range` qui permet de créer une séquence de nombres entiers. Elle fonctionne comme décrit ci-dessous
```python 
range(stop, start = 0, step = 1)
```
Les paramètres `start` et `step` peuvent être omis et ont les valeurs 0 et 1 respectivement. Notez également que la séquence d'entiers commencera à l'entier `start` inclusivement et se terminera à l'entier `stop` exclusivement, ce qui veut dire que la séquence d'entiers se terminera à `stop-1`.

Les boucles `while` sont similaires au boucle `for` à l'exception qu'elles vont continuer indéfiniment tant qu'une condition est satisfaite. On dit que la boucle `while` est une boucle conditionnelle (tandis que la boucle `for` est dite inconditionnelle)


En termes de syntaxe, la boucle while est utilisée comme suit
```python
while condition:
    #bloc de code à exécuter  
```

Comme vous le verrez dans la seconde série d'exercices, la boucle `while` sera notamment utile lorsqu'il faut répéter un algorithme indéfiniment jusqu'à ce qu'une condition sur une valeur critique ne soit plus respectée (voir exemple ci-dessous).

````python
epsilon_min = 1e-1

while epsilon_min > 1e-3:
    epsilon_min *= 1e-1
    print(epsilon_min)
````

Une boucle `while` peut remplacer une boucle `for` en ayant le même effet algorithmiquement. 

````python
i = 0
while i<10:
    print(i) 
    i+=1
````

Ce qui a le même effet que,

````python
for i in range(10):
    print(i)    
````

Astuces: Lorsque vous testez du code, vous pouvez sortir d'une boucle `while` qui tourne indéfiniment grâce à la commande Ctrl+C.

Il y a deux commandes utiles avec les boucles: la commande `break` et la commande `continue`. La commande `break` permet de sortir d'une boucle tandis que la commande `continue` permet de passer directement à l'itération suivante.  
Voyons deux exemples d'utilisation ci-dessous. 

````python
niter_max = 0 
epsilon = 10
while epsilon > 1e-5:
    niter_max +=1
    if niter_max > 1000:
        break
````

Ci-dessus, la commande `break` permet d'implémenter un nombre maximum d'itérations pour la boucle `while`. 

````python
sentence = "Ceci est une phrase test"
words = []
current_word = ""


for i in range(len(sentence)):
    current_letter = sentence[i]
    # print(current_letter)
    # If the current letter is a space, we have one word and we go to the next iteration
    if current_letter == " ":
        words.append(current_word)
        current_word=""
        continue
    current_word += current_letter
    if i == len(sentence)-1:
        words.append(current_word)

print(words)
````

Dans l'exemple ci-dessus, on veut séparer les différents mots dans la phrase en utilisant comme critère de séparation l'espace entre chaque mot. Le code ci-dessus reproduit donc le même effet que la méthode `split`

````python
sentence.split(" ")
````

### Les fonctions et les méthodes

On a rencontré quelques fonctions et méthodes durant le premier tutoriel et la première série d'exercices. Voici quelques exemples de fonction de base qui existe en Python et leur utilisation:
- `print()`: Affiche dans le terminal l'argument de la fonction
- `input()`: Demande à l'utilisateur du programme d'entrer une valeur.
- `len()`: Renvoie le nombre d'éléments d'une séquence


L'utilité principale des fonctions est de simplifier la vie du programmeur. Au lieu de réécrire plusieurs fois le même code pour faire tourner un algorithme qui ne change pas beaucoup, une fonction permet d'automatiser la tâche. Une autre utilité importante est qu'une fonction permet de rendre un programme plus lisible et permet de diviser les différentes étapes d'un programme complexe en sous-programmes plus simples.  

Ceci fait d'une fonction un bloc de code réutilisable permettant d'effectuer une tâche spécifique. 

Les fonctions se définissent grâce au mot `def` suivi par le nom de la fonction. Le bloc de code dans la fonction doit être indenté et la fonction peut retourner un résultat avec le mot `return`.
```python
def test_func(param, default_param = 0)
    # Écrire instructions code
    ...
    result = ...
    return result
```
La fonction prend ici en argument une variable `param` qui peut être utilisé dans le bloc de code et une variable `default_param`, un paramètre par défaut qui prend la valeur 0 s'il n'est pas spécifié par l'utilisateur. 

Testons ceci avec une fonction simple qui retourne un nombre au carré. 

````python
def square_func(x):
    return x ** 2 

print(square_func(2))
print(square_func(4))
````

On peut tirer un exemple de la première série d'exercices également avec l'exercice de la balle en chute libre lancée vers le haut et créer une fonction qui aura des paramètres par défaut. L'équation qu'on veut implémenter est la suivante 
$$
T = \frac{v_0 + \sqrt{v_0^2 + 2 g y_0}}{g}
$$
où $T$ est le temps de la chute libre et où $g$ est l'accélération de la pesanteur et $y_0$ l'altitude initiale de la balle.
On peut créer une fonction pour calculer la durée de vol de la balle. On pourrait coder la fonction de différentes manières mais ici je choisis d'avoir comme seul paramètres par défaut l'accélération de la pesanteur $g$ et les paramètres $v_0$ et $y_0$ d'être des paramètres obligatoires à spécifier pour utiliser la fonction.

````python
from math import sqrt

def flying_time(v0, y0, g = 9.81):
    T = (v0 + sqrt( v0 ** 2 + 2 * g * y0)) / g
    return T
````

Maintenant, on peut calculer très simplement la durée de vol de la balle selon un changement de vitesse initiale par exemple

````python
v0s = [10, 20, 30]
y0 = 10
for v0 in v0s:
    T = flying_time(v0, y0)
    print(f"For v0 = {v0:.2f}", f"Flying time = {T:.2f} sec")
````

On peut aussi faire le calcul selon un changement de l'accélération de la pesanteur. Par exemple, si le lancer s'effectue sur la Lune, on s'attend à ce que la durée de vol soit plus grande sur la Lune que sur Terre.

Astuce: La fonction `enumerate` utilisée sur une séquence renvoie une paire `(index, seq[index])`. Utilisé en combinaison avec une boucle `for`, `enumerate` permet d'itérer sur un entier en même temps qu'on itère sur la séquence. 

````python
gs = [9.81, 1.61, 3.82]
body_names = ["Earth", "Moon", "Mars"]

for i, g in enumerate(gs):
    T = flying_time(v0 = 10, y0 = 10, g = g)
    body = body_names[i]
    print("Flying time on", body, f"{T:.2f} sec")
````

Ceci vous fait également un bon exemple de la façon dont on peut utiliser les fonctions en combinaison avec des boucles. On verra à la prochaine séance comment on peut faire des opérations sur des tableaux avec le package `numpy` afin d'éviter les boucles itératives.

### Les erreurs

Il existe différents types d'erreurs en Python et il est bon de savoir les reconnaître afin d'être en capacité de chercher au bon endroit lorsqu'elles se produisent. Il existe:
- ***Les erreurs syntaxiques***: Si vous oubliez par exemple les `:` pour terminer une ligne où il y a l'instruction `for`. Ou si vous oubliez d'avoir un bloc d'indentation par exemple.
- ***Les erreurs d'exception***: Si vous demandez une opération qui est impossible à exécuter pour l'ordinateur. Vous verrez qu'il y a plusieurs sous-catégories d'erreurs d'exception, la division par zéro, l'utilisation de variables qui ne sont pas définies dans le code, ou encore des opérations impossibles pour des variables de type différents.

Finalement, les erreurs les plus vicieuses sont les erreurs abstraites. Pour celles-ci, votre code fonctionne parfaitement mais il exécute un bloc d'instructions menant à un résultat différent du résultat attendu. 


````python
for i in range(2)
    print("Hey !")
````
