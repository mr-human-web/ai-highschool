### Les différents types de variables en Python. 
Il y a:
- Les variables de type `str` (string). Elles correspondent au texte.
- Les variables de type `int`. Elles correspondent aux nombres entiers.
- Les variables de type `float`. Elles correspondent aux nombres réels.
- Les variables de type `list`. Elles correspondent à des listes dynamiques de variables  
- Les variables de type `tuple`. L'équivalent statique d'une liste.
- Les variables de type `dict`. Ce type de variable fonctionne comme un dictionnaire. 

Là où Python est très pratique par rapport aux autres langages de programmation c'est qu'il reconnait directement le type des variables.

Dans un autre langage par exemple, on aurait à écrire
```C
int a = 3;
```
pour définir un nombre entier alors qu'en Python, on a juste à écrire 
```python
a = 3
```

````python
# Example for the different variables in Python
str_var = "Hello"
int_var = 3
float_var = 3.14
list_var = [str_var, int_var, "Ender"]
tuple_var = (float_var, "Infosphere")
dict_var = {
    "int": int_var, 
    "list": list_var
}


# Afficher les différentes variables 
print(str_var,   "Type = ", type(str_var))
print(int_var,   "Type = ", type(int_var))
print(float_var, "Type = ", type(float_var))
print(tuple_var, "Type = ", type(tuple_var))
print(dict_var,  "Type = ", type(dict_var))
````

### Les variables de type int et float

Les variables de type `int` correspondent aux nombres entiers tandis que celles de types `float` correspondent au nombres réels.

Sans vouloir entrer trop dans des détails que je considère pas super intéressants, la différence entre ces deux types de variables est le nombre de bits utilisées afin de les encoder pour un ordinateur. 

Une variable de type int va en général utiliser 8 bits tandis qu'une variable de type float va en général utiliser 32 bits. 

En général, on n'a pas trop à se soucier de tout ça et on laisse l'ordinateur se soucier de ça... 

Commençons par créer deux variables, une de type `int` et une autre de type `float`

````python
var_int = 5 # integer variable
var_float = 3.14 # float variable

print("Type variable int", type(var_int))
print("Type variable float", type(var_float))
````

Nous pouvons effectuer des opérations avec ces variables. Voici certaines opérations et leur notation en programmation
- Multiplication -> `*`
- Division -> `/`
- Addition -> `+`
- Soustraction -> `-`
- Division entière -> `//`

Faisons un exemple pour chacune de ces opérations

````python
var1 = 11
var2 = 2

print("Multiplication", var1 * var2)
print("Division", var1 / var2)
print("Addition", var1 + var2)
print("Soustraction", var1 - var2)
print("Division entière", var1 // var2)
````

### Les variables de type `list`

Une variable de type `list` est une variable qui contient une liste dynamique d'éléments (qui peuvent être de type différents). Par dynamique, on entend ici qu'on peut ajouter/retirer des éléments à volonté dans une variable de type `list`.

On utilisera la méthode `.append()` pour ajouter des éléments à la fin d'une liste et `.remove()` pour supprimer des éléments d'une liste. Une autre méthode utile est `.insert(i, x)` qui permet d'ajouter un élément à une liste à une position précise. 

Quelque chose de très important est l'accès aux éléments composant une liste avec le format `var_list[idx]]` où `idx` est un nombre entier, un indice, indiquant la position de la variable qu'on souhaite accéder. Par exemple, si je veux accéder au premier élément d'une liste, je n'ai qu'à faire, 
```python
first_element = var_list[0]
```
(oui, en programmation on compte à partir de 0 pas à partir de 1).


````python
var_int = 5
var_list = [var_int, "c", 3]
var_list_bis = ["a", "b", "c"] 
print(var_list)
print(var_list + var_list_bis) # we can add lists together
print(var_list * 2)            # we can multiply a list by a int to repeat it
print(var_list_bis[0])         # we can access elements of a list. 
````

On peut aussi accéder un intervalle de valeurs en utilisant `:` (exemple ci-dessous)

````python
seq = var_list_bis[0:2] # accessing element 0 to element 1 (element 2 is excluded here)

var_list_long = [1, 2, 3, 4, 5, 6, 7]
seq_long = var_list_long[0:4:2] # from element to element 4 with a step of 2.

print("Séquence de la liste var_list_bis", seq)
print("Séquence de la liste var_list_long", seq_long)
````


### Les variables de type `dict`
On peut spécifier une variable de type `str` et lui associer une autre variable. Comme dans un dictionnaire où à chaque mot est associé une définition. On utilisera la notation suivante pour les variables de type `dict`:
```python
var_dict = {
    key_str_0: val_0,
    key_str_1: val_1
}
```
où `key_str_0` et `key_str_1` sont des variables de type string. 

````python
var_dict = {
    "mass": 15, # kg,
    "name": "Garfield",
    "age": 7, #years old
}


print(f"Le chat {var_dict['name']} a {var_dict['age']} ans et pèse {var_dict['mass']} kg")
print(type(var_dict))
````

### Les variables booléennes 

Les variables booléennes ne peuvent prendre que deux valeurs, soit `True` soit `False`. 

Voici un tableau d'opérations logiques pour ces variables: 
Opérateur & (and)
| (and) | True | False |
|:-----|----:|:-----:|
| True | True | False |
| False | False| False |

Opérateur | (or)
| (or) | True | False |
|:-----|----:|:-----:|
| True | True | True |
| False | True| False |

Ces variables nous sera notamment utile quand on écrira du code qui sera dépendant de condition. 

````python
print(True and False)
print(True & False) # une autre notation possible.
print(True or False)
print(True | False) # une autre notation possible. 
````

##### Chute libre
Une balle est lancée à la verticale vers le haut avec une vitesse initiale $v_0=10$ m/s. Nathan veut utiliser l'équation d'un objet en chute libre pour déterminer après combien de temps la balle retombera au sol (i.e. trouver l'instant $T$ où $y=0$ m).

L'équation d'un objet en chute libre est donnée par, 
$$
y = -\frac{1}{2}gt^2 + v_0 t + y_0 \, .
$$

Nathan déduit que la balle retombera au sol après une durée de temps $T$,
$$
-\frac{1}{2}gT^2 +v_0T +y_0 = 0
$$

Cette équation quadratique peut être résolue en utilisant la formule permettant d'en calculer les racines,
$$
\begin{align}
T_{\pm} &= \frac{-b\pm \sqrt{b^2 - 4ac}}{2 a}
\end{align}
$$
où
$$
\begin{align}
a &= -\frac{1}{2}g
\\
b &= v_0
\\
c&= y_0
\end{align} 
$$
On déduit donc que 
$$
\begin{align}
T_\pm &= \frac{- v_0\pm \sqrt{v_0^2+2gy_0}}{-g}
\end{align}
$$

````python
from math import sqrt
v0 = 30  # m/s
g = 9.81 # m/s^2
y0 = 2   # m

T = (-v0 - sqrt(v0 ** 2 + 2 * g * y0)) / -g
print(T, "secondes")
````
